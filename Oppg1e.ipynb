```python
#importerer bibloteker 
import numpy as np 
import math 
import matplotlib.pyplot as plt 

#definerer funkjsonen 
def sec(g, z0, z1, tol, max_iter = 100): 

    #lager en løkke med max interasjoner 
    for i in range(max_iter): 

        #definerer funkjsonsverdiene
        g_z0,g_z1 = g(z0), g(z1)

        #sjekker om nevneren i sekantmetoden ikke er for nærme null
        if abs(g_z1 - g_z0) < tol: 
            print("små forskjeller i funksjonsverdiene, kan gi mulig feil")
            break 

        # bruker sekantmetoden for å finne neste gjetning 
        z2 = z1 - g_z1*(z1 - z0)/(g_z1 - g_z0)

        #sjekker konvergens og hvis forskjellen er liten nokk returnerer vi z2: 
        if abs(z2 - z1) < tol: 
            return z2

        #oppdatere gjettene: 
        z0, z1 = z1, z2 

    raise ValueError("maks iterasjoner er nådd")

#vi definerer en funksjon, for å teste koden 
def g(z): 
    return z + math.sin(z) + math.cos(z)

#tester ut sekantmetoden 
z0 = -2
z1 = 0
tol = 1e-6

#finner roten 
root = sec(g, z0, z1, tol)
print(f"fant roten: {root}")

# sjekker om g(root) er nærme 0 
print(f"root function value: {g(root)}")
    

```

Ovenfor ser vi en funkjson som bruker sekantmetoden for å finne roten av funkjsonen g, deretter sjekker den om denne roten er riktig ved å sette roten inn for z i funksjonen g. Røttene til en funkjson, kaller vi også nullpunktene, siden det er punktene hvor g(z) = 0. Siden dette er en numerisk metode vil få en aproksimert rot, og g(root) vi ikke få eksakt 0, men en verdi som er veldig nærme 0.
