```python
import numpy as np 
import matplotlib.pyplot as plt 

def sec(g, z0, z1, tol, max_iter = 100): 

    for i in range(max_iter): 

        g_z0,g_z1 = g(z0), g(z1)
        
        if abs(g_z1 - g_z0) < tol: #unngår at divison med null 
            print("små forskjeller i funksjonsverdiene gjør ")
            break 

        z2 = z1 - g_z1*(z1 - z0)/(g_z1 - g_z0)

        #sjekker konvergens: 
        if abs(z2 - z1) < tol: 
            return z2

        #oppdatere gjettene: 
        z0, z1 = z1, z2 
    raise ValueError("")

#define the function g(z) = z+ sin(z) + cos(z)

import math 
def g(z): 
    return z + math.sin(z) + math.cos(z)

#test the secant method

z0 = -1
z1 = 0
tol = 1e-6

root = sec(g, z0, z1, tol)
print(f"root found: {root}")

#confim the root is correct
print(f"root function value: {g(root)}")
    

```
