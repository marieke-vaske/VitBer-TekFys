```python
import numpy as np
from scipy.integrate import solve_bvp
import matplotlib.pyplot as plt

def g_array(x, y_array):
    g_array = np.array([y_array[1], y_array[0] + np.sin(x)])
    return g_array
    
    
    
def bc(ya, yb):
    return np.array([ya[0], yb[0]])

x = np.linspace(0, 12, 100)  
y_guess = np.zeros((2, x.size))

solution = solve_bvp(g_array, bc,x, y_guess)


x_plot = np.linspace(0, np.pi/2, 100)
y_plot = solution.sol(x_plot)[0]  

plt.plot(solution.x, solution.y[0], label=r"$y(x)$", color="blue", linewidth=2)
plt.xlabel(r"$x$")
plt.ylabel(r"$y(x)$")
plt.title("Solution of the Boundary Value Problem")
plt.legend()
plt.grid()
plt.show()

# Print solution status
print("Success:", solution.success)
print("Message:", solution.message)
```
