```python
import numpy as np

def combineVectors(m_1, m_2, m_3, m_4):
    return np.concatenate((m_1, m_2, m_3, m_4))

# Test the combineVectors function
m_1 = np.array([1, 2, 3])
m_2 = np.array([4, 5, 6])
m_3 = np.array([7, 8, 9])
m_4 = np.array([10, 11, 12])

result = combineVectors(m_1, m_2, m_3, m_4)
print(result)

def seperateVectors(v):
    """  m_1, m_2, m_3, m_4 = np.array_split(v, 4) """
    return  np.array_split(v, 4)

print(seperateVectors(result))
    
```
