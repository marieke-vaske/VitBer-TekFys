# Oppgave 1b)

first we define the vector y(x): 

$$
\mathbf{y}(x) = 
\begin{bmatrix}
y(x) \\
y'(x)
\end{bmatrix}

$$


we define new variables: 
$$
y_1 = y , 

y_2 = y'
$$



$$
\frac{d}{dx}
\begin{bmatrix}
y_1 \\
y_2
\end{bmatrix}
=
\begin{bmatrix}
y_2 \\
-4 \sin(2x)
\end{bmatrix}, 
$$

initial condition: 

$$
\mathbf{y}(0) = 
\begin{bmatrix}
0\\
2
\end{bmatrix}
$$
